# Nepal Digital Transformation & Entrepreneurship — Panel Construction & Descriptive Analysis

Data: World Bank WDI export (`data.xlsx`), Nepal, 2006–2025, 13 indicators.

Steps:
1. Load raw WDI export
2. Reshape wide → long panel (Country-Year rows)
3. Clean missing values (`..` → NaN), drop empty 2025 column
4. Descriptive statistics
5. Charts: trends, correlation, scatter, distributions, growth index

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
from scipy.stats import zscore

plt.rcParams['font.family'] = 'DejaVu Sans'
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3

FILE = 'data.xlsx'   # <-- change to your local path if needed
```

## 1. Load raw data and inspect structure

```python
raw = pd.read_excel(FILE, sheet_name='Data')
print(raw.shape)
raw.head(15)
```

## 2. Reshape wide → long panel

Raw file has one row per **indicator**, one column per **year** (wide). We need one row per **Country-Year** (long/panel).

```python
# Drop footer/metadata rows that have no Series Code
raw = raw.dropna(subset=["Series Code"]).copy()

# Map WDI series codes to short variable names
code_map = {
    "IC.BUS.NDNS.ZS": "NBD",             # New business density
    "IT.NET.USER.ZS": "Internet",
    "IT.NET.BBND.P2": "Broadband",
    "IT.CEL.SETS.P2": "Mobile",
    "BX.GSR.CCIS.ZS": "ICT_Exports",
    "IT.NET.SECR.P6": "Secure_Servers",
    "NY.GDP.MKTP.KD.ZG": "GDP_Growth",
    "FS.AST.PRVT.GD.ZS": "Domestic_Credit",
    "BX.KLT.DINV.WD.GD.ZS": "FDI",
    "FP.CPI.TOTL.ZG": "Inflation",
    "EN.POP.DNST": "Pop_Density",
    "SE.SEC.ENRR": "School_Enroll",
    "SP.URB.TOTL.IN.ZS": "Urban_Pop"
}
raw["VarName"] = raw["Series Code"].map(code_map)

year_cols = [c for c in raw.columns if "YR" in c]
# Country identifiers are optional -- this extract may or may not include them
id_cols = [c for c in ["Country Name", "Country Code", "VarName"] if c in raw.columns]

long_df = raw[id_cols + year_cols].melt(id_vars=id_cols, var_name="YearRaw", value_name="Value")
long_df["Year"] = long_df["YearRaw"].str.extract(r"(\d{4})").astype(int)
long_df = long_df.drop(columns="YearRaw")

long_df.head()
```

## 3. Clean missing values and pivot into panel format

```python
# WDI encodes missing as ".." -> becomes NaN on numeric coercion
long_df["Value"] = pd.to_numeric(long_df["Value"], errors="coerce")

pivot_index = [c for c in ["Country Name", "Country Code", "Year"] if c in long_df.columns]
panel = long_df.pivot_table(index=pivot_index, columns="VarName", values="Value", aggfunc="first").reset_index()
panel.columns.name = None
panel = panel.sort_values("Year").reset_index(drop=True)

ordered = ["Country Name","Country Code","Year","NBD","Internet","Broadband","Mobile",
           "ICT_Exports","Secure_Servers","GDP_Growth","Domestic_Credit","FDI","Inflation",
           "Pop_Density","School_Enroll","Urban_Pop"]
panel = panel[[c for c in ordered if c in panel.columns]]

# Drop 2025 - essentially empty, not a real observation year
panel = panel[panel["Year"] <= 2024].reset_index(drop=True)

# Drop Country Name if present (single-country dataset, redundant) and make Year the row index
panel = panel.drop(columns=[c for c in ["Country Name"] if c in panel.columns])
panel = panel.set_index("Year")

print(panel.shape)
panel
```

```python
# Quick data-quality check
print("Missing values per variable:")
print(panel.isna().sum())

panel.to_csv('panel_data_clean.csv', index=True, index_label='Year')
```

## 4. Descriptive statistics

```python
analysis_vars = ['NBD','Internet','Broadband','Mobile','ICT_Exports','Secure_Servers',
                  'GDP_Growth','Domestic_Credit','FDI','Inflation','Pop_Density',
                  'School_Enroll','Urban_Pop']

labels = {
    'NBD':'New Business Density','Internet':'Internet Usage (%)',
    'Broadband':'Broadband Subs (/100)','Mobile':'Mobile Subs (/100)',
    'ICT_Exports':'ICT Exports (% services)','Secure_Servers':'Secure Servers (/1M)',
    'GDP_Growth':'GDP Growth (%)','Domestic_Credit':'Domestic Credit (% GDP)',
    'FDI':'FDI Inflows (% GDP)','Inflation':'Inflation (%)',
    'Pop_Density':'Population Density','School_Enroll':'School Enrollment (%)',
    'Urban_Pop':'Urban Population (%)'
}

desc = panel[analysis_vars].describe().T
desc['missing'] = panel[analysis_vars].isna().sum()
desc['skew'] = panel[analysis_vars].skew()
desc = desc.round(3)
desc
```

## 6. Why this is time-series, not panel data

This dataset has **one cross-sectional unit (Nepal)** observed over 19 years. With N=1, there is
no cross-sectional dimension to exploit — fixed/random effects, Hausman tests, and System GMM all
require N > 1. The correct toolkit for a single-country annual series like this is **time-series
econometrics**:

1. Unit root tests (ADF, Phillips-Perron) — determine order of integration
2. ARDL model — allows a mix of I(0) and I(1) regressors
3. Bounds test (Pesaran, Shin & Smith, 2001) — tests for long-run cointegration
4. Long-run coefficients (cointegrating vector)
5. Error Correction Model (ECM) — short-run dynamics + speed of adjustment
6. Diagnostic tests — serial correlation, heteroskedasticity, normality, functional form
7. CUSUM / CUSUMSQ — parameter stability over time

**Sample size caveat:** New Business Density (the entrepreneurship measure) is only available for
2006–2020 (15 annual observations). ARDL/bounds testing is normally applied to series of 30+
observations; with T=15 the results below are illustrative of the *methodology* rather than
statistically powerful. Given the degrees-of-freedom constraint, the working model here is kept
parsimonious: **NBD** (dependent) on **Internet usage** (primary digital-transformation regressor)
and **GDP growth** (macro control).

```python
from statsmodels.tsa.stattools import adfuller
from arch.unitroot import PhillipsPerron
from statsmodels.tsa.ardl import ARDL, ardl_select_order, UECM
from statsmodels.stats.diagnostic import (acorr_breusch_godfrey, het_breuschpagan,
                                           linear_reset, recursive_olsresiduals)
from statsmodels.stats.stattools import jarque_bera
import statsmodels.api as sm
import warnings
warnings.filterwarnings('ignore')

# Working sample: years where NBD is observed (2006-2020)
ts = panel.loc[2006:2020, ['NBD', 'Internet', 'GDP_Growth']].copy()
# ARDL needs a proper time index (not plain integers) to build lags correctly
ts.index = pd.period_range(start=str(ts.index[0]), periods=len(ts), freq='Y')
ts
```

## 7. ADF Unit Root Test

```python
def adf_test(series, name, maxlag=2):
    r = adfuller(series.dropna(), maxlag=maxlag, autolag='AIC')
    return {'Variable': name, 'ADF stat': round(r[0], 3), 'p-value': round(r[1], 3),
            'Lags used': r[2], 'Obs': r[3], 'Stationary (5%)': 'Yes' if r[1] < 0.05 else 'No'}

adf_rows = []
for col in ts.columns:
    adf_rows.append(adf_test(ts[col], f'{col} (level)'))
    adf_rows.append(adf_test(ts[col].diff(), f'{col} (1st diff)'))

adf_table = pd.DataFrame(adf_rows)
adf_table
```

## 8. Phillips-Perron Test

```python
def pp_test(series, name):
    pp = PhillipsPerron(series.dropna())
    return {'Variable': name, 'PP stat': round(pp.stat, 3), 'p-value': round(pp.pvalue, 3),
            'Stationary (5%)': 'Yes' if pp.pvalue < 0.05 else 'No'}

pp_rows = []
for col in ts.columns:
    pp_rows.append(pp_test(ts[col], f'{col} (level)'))
    pp_rows.append(pp_test(ts[col].diff(), f'{col} (1st diff)'))

pp_table = pd.DataFrame(pp_rows)
pp_table
```

**Reading the two tables together:** ADF and PP agree closely here. NBD and GDP growth look
stationary (or close to it) in levels — I(0); Internet usage is non-stationary in levels but
stationary after first-differencing — I(1). A mix of I(0) and I(1) regressors (with no I(2)
variables) is exactly the condition under which the **ARDL bounds testing** approach is valid —
this is why ARDL is a better fit here than a standard VAR/VECM, which would require every series
to be I(1).

## 9. ARDL Model

```python
# Fit ARDL(1,1,1): 1 lag of NBD, 1 lag of each exog regressor.
# (Note: with T=15, AIC-based automatic order selection can pick order 0 for an
# exog variable, which the UECM/bounds-test step below cannot use -- so lags are
# fixed here at 1 to keep the model estimable end-to-end.)
ardl_model = ARDL(ts["NBD"], lags=1, exog=ts[["Internet", "GDP_Growth"]], order=1, trend="c")
ardl_res = ardl_model.fit()
print("ARDL order (lags, exog order):", ardl_model.ardl_order)
print(ardl_res.summary())
```

## 10. Bounds Test for Cointegration

Pesaran, Shin & Smith (2001) F-bounds test, via the Unrestricted Error Correction Model (UECM)
representation of the ARDL. `case=3` = unrestricted intercept, no trend (matches `trend='c'` above).

**Decision rule:** if F-stat > upper bound → cointegration; if F-stat < lower bound → no
cointegration; if in between → inconclusive.

```python
uecm = UECM.from_ardl(ardl_model)
uecm_res = uecm.fit()
print(uecm_res.summary())

bounds = uecm_res.bounds_test(case=3)
print()
print(bounds)
```

## 11. Long-Run Estimation (Cointegrating Vector)

```python
long_run = uecm_res.ci_summary()
long_run
```

## 12. Error Correction Model

The UECM output above already contains the ECM: the coefficient on the lagged dependent variable
(`NBD.L1`) is the **speed-of-adjustment / error-correction term**. It should be negative and
between -1 and 0 for a stable long-run relationship — the fraction of last period's disequilibrium
that gets corrected each year. The `D.*` coefficients are the short-run dynamics.

```python
ec_coef = uecm_res.params['NBD.L1']
print(f'Error-correction coefficient (speed of adjustment): {ec_coef:.4f}')
if -1 < ec_coef < 0:
    half_life = np.log(0.5) / np.log(1 + ec_coef)
    print(f'Implied half-life of a shock: {half_life:.2f} years')
    print('Coefficient is negative and within (-1, 0): consistent with a stable, converging long-run relationship.')
else:
    print('Coefficient outside (-1, 0): no evidence of a stable error-correction process at this sample size.')
```

## 13. Diagnostic Tests

Run on the ARDL's underlying OLS design matrix (same fitted model, wrapped so statsmodels'
diagnostic functions can use it).

```python
X = ardl_model._x   # design matrix actually used in estimation (const + lags)
y = ardl_model._y
k = X.shape[1]
ols_res = sm.OLS(y, X).fit()   # numerically identical fit to ardl_res, wrapped for diagnostics

bg_stat, bg_p, _, _ = acorr_breusch_godfrey(ols_res, nlags=2)
jb_stat, jb_p, _, _ = jarque_bera(ols_res.resid)
bp_stat, bp_p, _, _ = het_breuschpagan(ols_res.resid, X)
reset = linear_reset(ols_res, power=2, use_f=True)

diag_table = pd.DataFrame([
    {'Test': 'Breusch-Godfrey (serial correlation)', 'Statistic': round(bg_stat,3), 'p-value': round(bg_p,3)},
    {'Test': 'Jarque-Bera (normality)', 'Statistic': round(jb_stat,3), 'p-value': round(jb_p,3)},
    {'Test': 'Breusch-Pagan (heteroskedasticity)', 'Statistic': round(bp_stat,3), 'p-value': round(bp_p,3)},
    {'Test': 'Ramsey RESET (functional form)', 'Statistic': round(float(reset.statistic),3), 'p-value': round(float(reset.pvalue),3)},
])
diag_table['Pass (p>0.05)'] = diag_table['p-value'] > 0.05
diag_table
```

## 14. CUSUM Test

Tests parameter stability using the cumulative sum of recursive residuals. The line should stay
inside the 5% critical bounds for the model to be judged stable.

```python
rresid, rparams, rypred, rresid_std, rresid_scaled, rcusum, _ = recursive_olsresiduals(
    ols_res, skip=k, alpha=0.95)

nrr = len(y) - k
a = 0.948  # 5% significance factor (Ploberger/Brown-Durbin-Evans)
bound = a * np.sqrt(nrr) + 2 * a * np.arange(0, nrr + 1) / np.sqrt(nrr)

fig, ax = plt.subplots(figsize=(9, 5))
xr = np.arange(len(rcusum))
ax.plot(xr, rcusum, marker='o', color='#2E86AB', label='CUSUM')
ax.plot(xr, bound, 'r--', label='5% critical bound')
ax.plot(xr, -bound, 'r--')
ax.axhline(0, color='gray', linewidth=0.8)
ax.set_title('CUSUM Test for Parameter Stability', fontsize=12, fontweight='bold')
ax.set_xlabel('Recursive step')
ax.legend()
plt.tight_layout()
plt.show()

stable = np.all(np.abs(rcusum) < bound)
print('Parameters stable over the sample (within 5% bounds):', stable)
```

## 15. CUSUMSQ Test

Tests stability of the residual variance using the cumulative sum of squared recursive residuals,
normalized to end at 1. Reference bounds shown here are the widely-used **exact bounds require
the published Brown-Durbin-Evans (1975) tables**; the diagonal line is the expected path under
parameter stability — visual departures from it flag instability.

```python
sq = rresid_std[k-1:] ** 2
cumsq = np.cumsum(sq) / np.sum(sq)
r_index = np.arange(1, len(cumsq) + 1)
expected = r_index / len(cumsq)

fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(r_index, cumsq, marker='o', color='#A23B72', label='CUSUMSQ')
ax.plot(r_index, expected, color='gray', linestyle=':', label='Expected path (stability)')
ax.set_title('CUSUMSQ Test for Parameter Stability', fontsize=12, fontweight='bold')
ax.set_xlabel('Recursive step')
ax.set_ylabel('Cumulative sum of squares (normalized)')
ax.legend()
plt.tight_layout()
plt.show()
```

## Summary & Limitations

- **Order of integration:** mixed I(0)/I(1) — ARDL bounds testing is the appropriate framework
  (confirmed by ADF/PP tests above).
- **Cointegration:** see the bounds-test F-statistic against the Pesaran/Shin/Smith critical
  values in Section 10.
- **Long-run relationship:** coefficients in Section 11.
- **Short-run adjustment:** error-correction speed in Section 12.
- **Model adequacy:** diagnostic tests in Section 13 — check none reject at 5%.
- **Stability:** CUSUM / CUSUMSQ in Sections 14-15.

**Key limitation:** T=15 (2006-2020) is well below the sample sizes typically used for reliable
ARDL bounds testing (usually 30+ annual observations recommended). Treat coefficient magnitudes
and p-values here as indicative, not confirmatory. If additional years of NBD data become
available (post-2020), or a longer pre-2006 series can be sourced, re-run this same pipeline with
the extended sample. The regressor set was deliberately kept to two variables (Internet, GDP
growth) to preserve degrees of freedom — this notebook's structure generalizes directly to
Broadband, Mobile, ICT_Exports, or Secure_Servers as alternative or additional regressors once a
larger sample supports them.
