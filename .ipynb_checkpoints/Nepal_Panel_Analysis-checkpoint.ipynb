# Nepal Digital Transformation & Entrepreneurship — Panel Construction & Descriptive Analysis

Data: World Bank WDI export (`data.xlsx`), Nepal, 2006–2025, 13 indicators.




```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
from scipy.stats import zscore

plt.rcParams['font.family'] = 'DejaVu Sans'
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3

FILE = 'data.xlsx'  
```

## 1. Load raw data and inspect structure

```python
raw = pd.read_excel(FILE, sheet_name='Data')
print(raw.shape)
raw.head(15)
```

## 2. Reshape wide → long panel

Raw file has one row per **indicator**, one column per **year** (wide). We need one row per **Country-Year** (long/panel).

```python
# Drop footer/metadata rows that have no Series Code
raw = raw.dropna(subset=['Series Code']).copy()

# Map WDI series codes to short variable names
code_map = {
    'IC.BUS.NDNS.ZS': 'NBD',             # New business density
    'IT.NET.USER.ZS': 'Internet',
    'IT.NET.BBND.P2': 'Broadband',
    'IT.CEL.SETS.P2': 'Mobile',
    'BX.GSR.CCIS.ZS': 'ICT_Exports',
    'IT.NET.SECR.P6': 'Secure_Servers',
    'NY.GDP.MKTP.KD.ZG': 'GDP_Growth',
    'FS.AST.PRVT.GD.ZS': 'Domestic_Credit',
    'BX.KLT.DINV.WD.GD.ZS': 'FDI',
    'FP.CPI.TOTL.ZG': 'Inflation',
    'EN.POP.DNST': 'Pop_Density',
    'SE.SEC.ENRR': 'School_Enroll',
    'SP.URB.TOTL.IN.ZS': 'Urban_Pop'
}
raw['VarName'] = raw['Series Code'].map(code_map)

year_cols = [c for c in raw.columns if 'YR' in c]
id_cols = ['Country Name', 'Country Code', 'VarName']

long_df = raw[id_cols + year_cols].melt(id_vars=id_cols, var_name='YearRaw', value_name='Value')
long_df['Year'] = long_df['YearRaw'].str.extract(r'(\d{4})').astype(int)
long_df = long_df.drop(columns='YearRaw')

long_df.head()
```

## 3. Clean missing values and pivot into panel format

```python
# WDI encodes missing as '..' -> becomes NaN on numeric coercion
long_df["Value"] = pd.to_numeric(long_df["Value"], errors="coerce")

panel = long_df.pivot_table(index=["Country Name", "Country Code", "Year"],
                             columns="VarName", values="Value", aggfunc="first").reset_index()
panel.columns.name = None
panel = panel.sort_values("Year").reset_index(drop=True)

ordered = ["Country Name","Country Code","Year","NBD","Internet","Broadband","Mobile",
           "ICT_Exports","Secure_Servers","GDP_Growth","Domestic_Credit","FDI","Inflation",
           "Pop_Density","School_Enroll","Urban_Pop"]
panel = panel[[c for c in ordered if c in panel.columns]]

# Drop 2025 - essentially empty, not a real observation year
panel = panel[panel["Year"] <= 2024].reset_index(drop=True)

# Drop Country Name (single-country dataset, redundant) and make Year the row index
panel = panel.drop(columns=["Country Name"])
panel = panel.set_index("Year")

print(panel.shape)
panel
```

```python
# Quick data-quality check
print("Missing values per variable:")
print(panel.isna().sum())

panel.to_csv('panel_data_clean.csv', index=True, index_label='Year')
```

## 4. Descriptive statistics

```python
analysis_vars = ['NBD','Internet','Broadband','Mobile','ICT_Exports','Secure_Servers',
                  'GDP_Growth','Domestic_Credit','FDI','Inflation','Pop_Density',
                  'School_Enroll','Urban_Pop']

labels = {
    'NBD':'New Business Density','Internet':'Internet Usage (%)',
    'Broadband':'Broadband Subs (/100)','Mobile':'Mobile Subs (/100)',
    'ICT_Exports':'ICT Exports (% services)','Secure_Servers':'Secure Servers (/1M)',
    'GDP_Growth':'GDP Growth (%)','Domestic_Credit':'Domestic Credit (% GDP)',
    'FDI':'FDI Inflows (% GDP)','Inflation':'Inflation (%)',
    'Pop_Density':'Population Density','School_Enroll':'School Enrollment (%)',
    'Urban_Pop':'Urban Population (%)'
}

desc = panel[analysis_vars].describe().T
desc['missing'] = panel[analysis_vars].isna().sum()
desc['skew'] = panel[analysis_vars].skew()
desc = desc.round(3)
desc
```

## 5. Charts

### 5.1 Digital transformation & entrepreneurship trends

```python
dt_vars = ['NBD','Internet','Broadband','Mobile','ICT_Exports','Secure_Servers']
colors = ['#2E86AB','#A23B72','#F18F01','#C73E1D','#3B1F2B','#6A994E']

fig, axes = plt.subplots(3, 2, figsize=(14, 12))
for ax, v, c in zip(axes.flat, dt_vars, colors):
    d = panel[v].dropna()
    ax.plot(d.index, d.values, marker='o', color=c, linewidth=2, markersize=4)
    ax.set_title(labels[v], fontsize=11, fontweight='bold')
    ax.set_xlabel('Year')
    ax.xaxis.set_major_locator(mticker.MaxNLocator(integer=True, nbins=8))
fig.suptitle('Nepal: Digital Transformation & Entrepreneurship Trends (2006-2024)', fontsize=14, fontweight='bold')
plt.tight_layout(rect=[0,0,1,0.96])
plt.show()
```

### 5.2 Macroeconomic moderating variables

```python
macro_vars = ['GDP_Growth','Domestic_Credit','FDI','Inflation']
colors2 = ['#264653','#2A9D8F','#E76F51','#E9C46A']

fig, axes = plt.subplots(2, 2, figsize=(14, 9))
for ax, v, c in zip(axes.flat, macro_vars, colors2):
    d = panel[v].dropna()
    ax.plot(d.index, d.values, marker='s', color=c, linewidth=2, markersize=4)
    ax.axhline(0, color='gray', linewidth=0.8, linestyle='--')
    ax.set_title(labels[v], fontsize=11, fontweight='bold')
    ax.set_xlabel('Year')
    ax.xaxis.set_major_locator(mticker.MaxNLocator(integer=True, nbins=8))
fig.suptitle('Nepal: Macroeconomic Moderating Variables (2006-2024)', fontsize=14, fontweight='bold')
plt.tight_layout(rect=[0,0,1,0.96])
plt.show()
```

### 5.3 Correlation heatmap

```python
corr = panel[analysis_vars].corr()

fig, ax = plt.subplots(figsize=(11,9))
im = ax.imshow(corr, cmap='RdBu_r', vmin=-1, vmax=1)
ax.set_xticks(range(len(analysis_vars)))
ax.set_yticks(range(len(analysis_vars)))
ax.set_xticklabels([labels[v] for v in analysis_vars], rotation=45, ha='right', fontsize=9)
ax.set_yticklabels([labels[v] for v in analysis_vars], fontsize=9)
for i in range(len(analysis_vars)):
    for j in range(len(analysis_vars)):
        val = corr.iloc[i,j]
        ax.text(j, i, f'{val:.2f}', ha='center', va='center',
                color='white' if abs(val)>0.5 else 'black', fontsize=8)
plt.colorbar(im, label='Pearson correlation')
ax.set_title('Correlation Matrix', fontsize=12, fontweight='bold')
plt.tight_layout()
plt.show()
```

### 5.4 New Business Density vs. each digital transformation dimension

```python
dt_predictors = ['Internet','Broadband','Mobile','ICT_Exports','Secure_Servers']

fig, axes = plt.subplots(2, 3, figsize=(15, 9))
for ax, v, c in zip(axes.flat, dt_predictors, colors[1:]):
    d = panel[['NBD', v]].dropna()
    ax.scatter(d[v], d['NBD'], color=c, alpha=0.7, edgecolor='white', s=60)
    if len(d) > 2:
        z = np.polyfit(d[v], d['NBD'], 1)
        xs = np.linspace(d[v].min(), d[v].max(), 50)
        ax.plot(xs, np.polyval(z, xs), color='black', linewidth=1.5, linestyle='--')
        r = d[v].corr(d['NBD'])
        ax.text(0.05, 0.92, f'r = {r:.2f}', transform=ax.transAxes, fontsize=10, fontweight='bold')
    ax.set_xlabel(labels[v], fontsize=9)
    ax.set_ylabel('New Business Density', fontsize=9)
axes.flat[-1].axis('off')
fig.suptitle('New Business Density vs. Digital Transformation Dimensions', fontsize=14, fontweight='bold')
plt.tight_layout(rect=[0,0,1,0.95])
plt.show()
```

### 5.5 Distribution / outlier check (standardized)

```python
z_df = panel[dt_vars].apply(lambda x: zscore(x, nan_policy='omit'))

fig, ax = plt.subplots(figsize=(10,6))
z_df.boxplot(ax=ax)
ax.set_xticklabels([labels[v] for v in dt_vars], rotation=30, ha='right')
ax.set_ylabel('Standardized value (z-score)')
ax.set_title('Distribution & Outlier Check', fontsize=12, fontweight='bold')
ax.axhline(0, color='gray', linewidth=0.8)
plt.tight_layout()
plt.show()
```

### 5.6 Growth index (2010 = 100)

```python
base_year = 2010
idx_df = panel[dt_vars]
idx_df = idx_df / idx_df.loc[base_year] * 100

fig, ax = plt.subplots(figsize=(11,6.5))
for v, c in zip(dt_vars, colors):
    ax.plot(idx_df.index, idx_df[v], marker='o', markersize=3, label=labels[v], color=c)
ax.axhline(100, color='gray', linewidth=0.8, linestyle=':')
ax.set_ylabel(f'Index ({base_year} = 100)')
ax.set_xlabel('Year')
ax.set_title('Growth Trajectories: Digital Transformation & Entrepreneurship (2010=100)', fontsize=12, fontweight='bold')
ax.legend(loc='upper left', fontsize=8, ncol=2)
plt.tight_layout()
plt.show()
```
