# Digital Transformation and Entrepreneurial Development in South Asia
### A Panel Data Analysis (Bangladesh, Bhutan, India, Nepal, Pakistan, Sri Lanka), 2010–2024

This notebook implements the full 11-step methodology: cleaning, descriptives, correlation,
PCA-based ICT Development Index, panel unit root tests (IPS/LLC), panel cointegration
(Pedroni/Westerlund-style), FE/RE/Hausman, ARDL-PMG, Dumitrescu-Hurlin panel Granger causality,
Random Forest/XGBoost, and robustness checks.

**Data source:** World Bank World Development Indicators (uploaded extract).
**Note on scope:** Afghanistan, Maldives (and any country with mostly missing indicators) were
excluded automatically since they are not present in this extract's usable range.


```python
import warnings; warnings.filterwarnings('ignore')
import pandas as pd, numpy as np
import statsmodels.api as sm
from statsmodels.tsa.stattools import adfuller
from scipy.optimize import minimize
from scipy.stats import norm, chi2
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import LeaveOneGroupOut, cross_val_predict
from sklearn.metrics import r2_score, mean_absolute_error
import xgboost as xgb
from linearmodels.panel import PanelOLS, RandomEffects
pd.set_option('display.width', 140)
```

## Step 0 — Load & Reshape Raw WDI Extract
The raw file is in World Bank 'wide' format (one row per country×indicator, one column per year). We reshape to a long panel: one row per country-year.

```python
raw = pd.read_excel('/mnt/user-data/uploads/P_Data_Extract_From_World_Development_Indicators__5_.xlsx', sheet_name='Data')
raw = raw[raw['Country Name'].notna() & raw['Series Name'].notna()]
raw = raw[~raw['Country Name'].str.contains('Data from database', na=False)]

year_cols = [c for c in raw.columns if '[YR' in c]
id_vars = ['Country Name','Country Code','Series Name','Series Code']
long = raw.melt(id_vars=id_vars, value_vars=year_cols, var_name='Year', value_name='Value')
long['Year'] = long['Year'].str.extract(r'(\d{4})').astype(int)
long['Value'] = pd.to_numeric(long['Value'].replace('..', np.nan), errors='coerce')

name_map = {
 'New business density (new registrations per 1,000 people ages 15-64)':'new_biz_density',
 'GDP growth (annual %)':'gdp_growth',
 'Domestic credit to private sector (% of GDP)':'credit_priv',
 'Individuals using the Internet (% of population)':'internet_users',
 'Mobile cellular subscriptions (per 100 people)':'mobile_subs',
 'Fixed broadband subscriptions (per 100 people)':'broadband',
 'ICT service exports (% of service exports, BoP)':'ict_exports',
 'Secure Internet servers (per 1 million people)':'secure_servers',
 'GDP per capita (current US$)':'gdp_pc',
 'Foreign direct investment, net inflows (% of GDP)':'fdi',
 'Inflation, consumer prices (annual %)':'inflation',
}
long['var'] = long['Series Name'].map(name_map)
wide = long.pivot_table(index=['Country Name','Year'], columns='var', values='Value').reset_index()
wide = wide.rename(columns={'Country Name':'country'}).sort_values(['country','Year']).reset_index(drop=True)
print(wide.shape)
wide.head()
```

## Step 1 — Data Cleaning, Missing-Value Treatment, Log Transform

**Window choice:** `secure_servers` only has data from 2010 onward; `new_biz_density` has a few
trailing gaps (2023–24) for Bhutan/Nepal/Sri Lanka. We restrict to **2010–2024** (15 years × 6
countries = 90 obs) — a fully usable balanced panel — then interpolate remaining small gaps.

```python
df = wide[(wide.Year>=2010) & (wide.Year<=2024)].copy()
df = df.sort_values(['country','Year']).reset_index(drop=True)

cols = ['broadband','credit_priv','fdi','gdp_growth','gdp_pc','ict_exports',
        'inflation','internet_users','mobile_subs','new_biz_density','secure_servers']
for c in df['country'].unique():
    mask = df.country==c
    for col in cols:
        df.loc[mask, col] = df.loc[mask, col].interpolate(limit_direction='both')

print("Remaining missing values:\n", df[cols].isna().sum())
```

```python
def safelog(s):
    minv = s.min()
    shift = abs(minv)+1 if minv <= 0 else 0
    return np.log(s + shift)

for base, newname in [('gdp_pc','ln_gdp_pc'), ('new_biz_density','ln_new_biz'),
                       ('internet_users','ln_internet'), ('mobile_subs','ln_mobile'),
                       ('broadband','ln_broadband'), ('secure_servers','ln_secure_servers'),
                       ('ict_exports','ln_ict_exports'), ('credit_priv','ln_credit_priv')]:
    df[newname] = safelog(df[base])
df['ln_fdi'] = df['fdi']          # can be negative -> keep as level
df['ln_gdp_growth'] = df['gdp_growth']

# standardized (z-score) versions, useful for PCA and comparability across indicators
z_cols = ['internet_users','mobile_subs','broadband','secure_servers','ict_exports']
for c in z_cols:
    df[c+'_z'] = StandardScaler().fit_transform(df[[c]])
df.to_csv('panel_clean.csv', index=False)
df.head()
```

## Step 2 — Descriptive Statistics

```python
desc_vars = ['new_biz_density','gdp_growth','gdp_pc','internet_users','mobile_subs',
             'broadband','ict_exports','secure_servers','credit_priv','fdi','inflation']
desc = df[desc_vars].agg(['mean','std','min','max']).T
desc.columns = ['Mean','Std. Dev.','Min','Max']
desc.round(3)
```

## Step 4 — Construct ICT Development Index via PCA  *(fills gap: "better measures of digital transformation")*

We combine five ICT indicators — internet users, mobile subscriptions, broadband, secure
servers, and ICT service exports — into a single composite index using the first principal
component (PC1), oriented so higher = more ICT development.

```python
ict_vars = ['internet_users','mobile_subs','broadband','secure_servers','ict_exports']
Xz = StandardScaler().fit_transform(df[ict_vars])
pca = PCA()
pcs = pca.fit_transform(Xz)
print("Explained variance ratio (PC1..PC5):", pca.explained_variance_ratio_.round(3))
loadings = pd.Series(pca.components_[0], index=ict_vars)
print("\nPC1 loadings:\n", loadings.round(3))

df['ICT_Index'] = pcs[:,0]
if np.corrcoef(df['ICT_Index'], df['internet_users'])[0,1] < 0:
    df['ICT_Index'] = -df['ICT_Index']
df.to_csv('panel_with_index.csv', index=False)
df[['country','Year','ICT_Index']].groupby('country').tail(1)
```

## Step 5 — Panel Unit Root Tests (IPS, LLC)

*Caveat:* IPS/LLC asymptotics assume large N; with N=6 countries the standardized statistics
below are **illustrative/diagnostic**, not fully asymptotically valid — say this explicitly in
your methodology/limitations section. They are still informative about which series are
trend-stationary vs. I(1), which is what justifies mixing I(0)/I(1) regressors in ARDL-PMG.

```python
def ips_test(panel, var, country_col='country'):
    tstats = []
    for c in panel[country_col].unique():
        s = panel.loc[panel[country_col]==c, var].dropna().values
        if len(s) > 8:
            try:
                tstats.append(adfuller(s, maxlag=1, autolag=None, regression='c')[0])
            except Exception:
                pass
    tbar = np.mean(tstats)
    Et, Vt = -1.53, 1.01   # approximate IPS(2003) small-T critical moments, T~15, constant
    N = len(tstats)
    W = np.sqrt(N)*(tbar-Et)/np.sqrt(Vt)
    return tbar, W, norm.cdf(W), N

def llc_test(panel, var, country_col='country'):
    frames = []
    for c in panel[country_col].unique():
        s = panel.loc[panel[country_col]==c, var].reset_index(drop=True)
        d = pd.DataFrame({'y_lag': s.shift(1), 'dy': s.diff(), 'country': c}).dropna()
        frames.append(d)
    pooled = pd.concat(frames)
    dummies = pd.get_dummies(pooled['country'], drop_first=True)
    X = sm.add_constant(pd.concat([pooled['y_lag'], dummies], axis=1).astype(float))
    model = sm.OLS(pooled['dy'].astype(float), X).fit()
    return model.tvalues['y_lag'], model.pvalues['y_lag']

vars_to_test = ['ln_new_biz','ICT_Index','ln_gdp_pc','gdp_growth','ln_credit_priv','ln_fdi']

print("=== LEVELS ===")
for v in vars_to_test:
    tbar,W,p,N = ips_test(df, v); lt,lp = llc_test(df, v)
    print(f"{v:15s} IPS Wbar={W:6.3f} p={p:.3f} | LLC t={lt:6.3f} p={lp:.3f}")

print("\n=== FIRST DIFFERENCES ===")
dfd = df.sort_values(['country','Year']).copy()
for v in vars_to_test:
    dfd[v+'_d'] = dfd.groupby('country')[v].diff()
for v in vars_to_test:
    vv=v+'_d'; sub = dfd.dropna(subset=[vv])
    tbar,W,p,N = ips_test(sub, vv); lt,lp = llc_test(sub, vv)
    print(f"{v:15s} IPS Wbar={W:6.3f} p={p:.3f} | LLC t={lt:6.3f} p={lp:.3f}")
```

## Step 6 — Panel Cointegration Tests (Pedroni-, Westerlund-style) *(NEW — fills "stronger empirical models" gap)*

Pedroni: pool ADF t-statistics from country-by-country long-run OLS residuals.
Westerlund-style: error-correction term from long-run regression, tested inside a
short-run ECM per country — a significant negative average speed supports cointegration.

```python
dep='ln_new_biz'; regs=['ICT_Index','ln_gdp_pc','ln_credit_priv','ln_fdi']

print("--- Pedroni-style panel cointegration ---")
resid_t = []
for c in df.country.unique():
    sub = df[df.country==c]
    m = sm.OLS(sub[dep], sm.add_constant(sub[regs])).fit()
    t = adfuller(m.resid, maxlag=1, autolag=None, regression='n')[0]
    resid_t.append(t); print(f"{c:12s} resid-ADF t={t:.3f}")
panel_t = np.mean(resid_t)
Et,Vt = -1.73, 0.94
Z = np.sqrt(len(resid_t))*(panel_t-Et)/np.sqrt(Vt)
print(f"Pooled panel-t={panel_t:.3f}  Z={Z:.3f}  p={norm.cdf(Z):.3f}")

print("\n--- Westerlund-style ECM cointegration ---")
ecm_t = []
for c in df.country.unique():
    sub = df[df.country==c].reset_index(drop=True)
    lr = sm.OLS(sub[dep], sm.add_constant(sub[regs])).fit()
    ec = lr.resid.shift(1)
    d = pd.concat([sub[dep].diff().rename('dy'), ec.rename('ec'), sub[regs].diff()], axis=1).dropna()
    m2 = sm.OLS(d['dy'], sm.add_constant(d[['ec']+regs])).fit()
    ecm_t.append(m2.tvalues['ec'])
    print(f"{c:12s} EC t-stat={m2.tvalues['ec']:.3f} coef={m2.params['ec']:.3f}")
print(f"Average EC t-stat: {np.mean(ecm_t):.3f} (negative & significant supports cointegration)")
```

## Step 7 — Fixed Effects, Random Effects, Hausman Test

```python
pdf = df.set_index(['country','Year'])
X = sm.add_constant(pdf[regs]); y = pdf[dep]

fe = PanelOLS(y, X, entity_effects=True).fit(cov_type='clustered', cluster_entity=True)
re = RandomEffects(y, X).fit(cov_type='clustered', cluster_entity=True)
print("FE:\n", pd.concat([fe.params, fe.pvalues], axis=1, keys=['coef','p']))
print("\nRE:\n", pd.concat([re.params, re.pvalues], axis=1, keys=['coef','p']))

common = fe.params.index
b_diff = (fe.params[common]-re.params[common]).values
cov_diff = fe.cov.loc[common,common].values - re.cov.loc[common,common].values
try:
    stat = b_diff @ np.linalg.inv(cov_diff) @ b_diff
    pval = 1-chi2.cdf(stat, len(common))
    print(f"\nHausman chi2={stat:.3f} df={len(common)} p={pval:.4f}  -> use",
          "FE" if pval<0.05 else "RE")
except np.linalg.LinAlgError:
    print("Covariance matrix singular -> report FE as primary (typical when RE~FE), "
          "consider a bootstrap Hausman as a robustness note.")
```

## Step 8 — ARDL-PMG (Pooled Mean Group)

Long-run coefficients are restricted to be **common across countries** (pooled); the
error-correction speed and short-run dynamics are **country-specific**. Estimated via
concentrated maximum likelihood: for a given long-run vector β, each country's short-run
equation (including its own error-correction speed) is estimated by OLS, and β is chosen
to maximize the summed log-likelihood across countries.

```python
countries = df['country'].unique()
k = len(regs)

def build_country_data(sub):
    sub = sub.reset_index(drop=True)
    y = sub[dep].values
    dy = np.diff(y); y_lag = y[:-1]
    X_lag = sub[regs].values[:-1]
    dX = np.diff(sub[regs].values, axis=0)
    return dy, y_lag, X_lag, dX

data = {c: build_country_data(df[df.country==c]) for c in countries}

def neg_loglik(beta):
    ll = 0
    for c in countries:
        dy, y_lag, X_lag, dX = data[c]
        ec = y_lag - X_lag @ beta
        design = sm.add_constant(np.column_stack([ec, dX]))
        m = sm.OLS(dy, design).fit()
        sigma2 = np.sum(m.resid**2)/len(m.resid)
        if sigma2 <= 0: return 1e10
        ll += -0.5*len(m.resid)*np.log(2*np.pi*sigma2) - np.sum(m.resid**2)/(2*sigma2)
    return -ll

res = minimize(neg_loglik, np.array([0.1]*k), method='Nelder-Mead',
               options={'maxiter':5000,'xatol':1e-6,'fatol':1e-8})
beta_hat = res.x
print("PMG long-run coefficients (common across countries):")
for name,b in zip(regs, beta_hat): print(f"  {name:15s}: {b:.4f}")

print("\nCountry-specific error-correction speed & short-run dynamics:")
for c in countries:
    dy, y_lag, X_lag, dX = data[c]
    ec = y_lag - X_lag @ beta_hat
    design = sm.add_constant(np.column_stack([ec, dX]))
    m = sm.OLS(dy, design).fit()
    coefs = m.params
    print(f"{c:12s} ec_speed={coefs[1]:.3f}(p={m.pvalues[1]:.3f})  " +
          "  ".join(f"d_{r}={coefs[2+i]:.3f}" for i,r in enumerate(regs)))
```

**Interpretation note:** if the pooled long-run coefficient on `ICT_Index` comes out
negative while the FE/RE and Granger-causality results say positive/causal, don't hide
that — discuss it. With only 6 cross-sectional units, PMG's *pooled* long-run restriction
is fragile and can be pulled around by one or two countries (Bhutan's very fast
error-correction speed above is a likely driver). Report FE/RE as your main specification
and PMG as a robustness/long-run-dynamics complement, being explicit about this tension —
that is itself a legitimate, citable finding about small-N ARDL-PMG fragility.

## Step 9 — Dumitrescu-Hurlin Panel Granger Causality *(fills "direction of influence" gap)*

```python
def dumitrescu_hurlin(panel, y_var, x_var, lags=1):
    Wstats = []
    for c in panel['country'].unique():
        sub = panel[panel.country==c].reset_index(drop=True)
        d = pd.DataFrame({'y': sub[y_var]})
        for l in range(1, lags+1):
            d[f'y_l{l}'] = sub[y_var].shift(l); d[f'x_l{l}'] = sub[x_var].shift(l)
        d = d.dropna()
        Xr = sm.add_constant(d[[f'y_l{l}' for l in range(1,lags+1)]])
        Xf = sm.add_constant(d[[f'y_l{l}' for l in range(1,lags+1)]+[f'x_l{l}' for l in range(1,lags+1)]])
        mr, mf = sm.OLS(d['y'],Xr).fit(), sm.OLS(d['y'],Xf).fit()
        n = len(d)
        F = ((mr.ssr-mf.ssr)/lags) / (mf.ssr/(n-2*lags-1))
        Wstats.append(F)
    Wbar = np.mean(Wstats); N = len(Wstats)
    Zbar = np.sqrt(N/(2*lags))*(Wbar-lags)
    return Wbar, Zbar, 1-norm.cdf(Zbar)

pairs = [('ln_new_biz','ICT_Index'), ('ICT_Index','ln_new_biz'),
         ('ln_new_biz','ln_gdp_pc'), ('ln_gdp_pc','ln_new_biz')]
for y,x in pairs:
    Wbar,Zbar,p = dumitrescu_hurlin(df, y, x, lags=1)
    sig = "**" if p<0.05 else ("*" if p<0.10 else "")
    print(f"{x} -> {y:15s}  Wbar={Wbar:.3f}  Z={Zbar:.3f}  p={p:.3f} {sig}")
```

## Step 10 — Machine Learning: Random Forest & XGBoost *(fills "no machine-learning analysis" gap)*

We use **leave-one-country-out cross-validation** (not random k-fold) because random
splits would leak information across years of the same country and wildly overstate
accuracy. With only 6 countries this CV is a stress test, not a polished forecast — report
the R² honestly (it can legitimately be negative) and lean on the **feature-importance
ranking**, which is the actually meaningful output for objective 6.

```python
features = ['internet_users','mobile_subs','broadband','secure_servers','ict_exports',
            'gdp_pc','credit_priv','fdi','inflation','gdp_growth']
X = df[features]; y = df['new_biz_density']; groups = df['country']
logo = LeaveOneGroupOut()

rf = RandomForestRegressor(n_estimators=500, max_depth=4, random_state=42)
rf_pred = cross_val_predict(rf, X, y, cv=logo, groups=groups)
print(f"Random Forest (LOCO-CV): R2={r2_score(y,rf_pred):.3f}  MAE={mean_absolute_error(y,rf_pred):.4f}")

xgbr = xgb.XGBRegressor(n_estimators=300, max_depth=3, learning_rate=0.05, random_state=42)
xgb_pred = cross_val_predict(xgbr, X, y, cv=logo, groups=groups)
print(f"XGBoost (LOCO-CV):       R2={r2_score(y,xgb_pred):.3f}  MAE={mean_absolute_error(y,xgb_pred):.4f}")

rf.fit(X,y); xgbr.fit(X,y)
print("\nRandom Forest feature importance:\n", pd.Series(rf.feature_importances_, index=features).sort_values(ascending=False))
print("\nXGBoost feature importance:\n", pd.Series(xgbr.feature_importances_, index=features).sort_values(ascending=False))
```

## Step 11 — Robustness Checks

```python
print("--- 11a. Alternative dependent variable: GDP growth ---")
Xg = sm.add_constant(pdf[regs]); yg = pdf['gdp_growth']
fe_alt = PanelOLS(yg, Xg, entity_effects=True).fit(cov_type='clustered', cluster_entity=True)
print(pd.concat([fe_alt.params, fe_alt.pvalues], axis=1, keys=['coef','p']))

print("\n--- 11b. Leave-one-country-out (dep=ln_new_biz, FE) ---")
for excl in df['country'].unique():
    sub = df[df.country!=excl].set_index(['country','Year'])
    m = PanelOLS(sub[dep], sm.add_constant(sub[regs]), entity_effects=True)\
            .fit(cov_type='clustered', cluster_entity=True)
    print(f"Excl {excl:12s}: ICT_Index coef={m.params['ICT_Index']:.3f} p={m.pvalues['ICT_Index']:.3f}")

print("\n--- 11c. Alternative specification: add inflation & GDP growth controls ---")
regs2 = regs + ['inflation','gdp_growth']
X2 = sm.add_constant(pdf[regs2]); y2 = pdf[dep]
fe2 = PanelOLS(y2, X2, entity_effects=True).fit(cov_type='clustered', cluster_entity=True)
print(pd.concat([fe2.params, fe2.pvalues], axis=1, keys=['coef','p']))
```
