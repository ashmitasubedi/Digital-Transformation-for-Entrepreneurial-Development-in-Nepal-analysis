# Digital Transformation and Entrepreneurial Development in South Asia
### Panel Econometric Analysis (2010–2023)

**Data:** World Bank World Development Indicators (WDI), six South Asian countries — Nepal, India, Bangladesh, Pakistan, Sri Lanka, Bhutan.

**Dependent variable:** New business density (`NBD`)
**Digital transformation variables:** Internet users (`INTERNET`), Mobile subscriptions (`MOBILE`), Fixed broadband (`BBAND`), Secure Internet servers (`SECSERV`)
**Controls:** GDP per capita (`GDPPC`), Domestic credit to private sector (`CREDIT`), FDI net inflows (`FDI`), Inflation (`INFLATION`)

**Model:**
$$NBD_{it} = \beta_0 + \beta_1 INTERNET_{it} + \beta_2 MOBILE_{it} + \beta_3 BBAND_{it} + \beta_4 SECSERV_{it} + \beta_5 GDPPC_{it} + \beta_6 CREDIT_{it} + \beta_7 FDI_{it} + \beta_8 INFLATION_{it} + \mu_i + \lambda_t + \varepsilon_{it}$$

This notebook follows the 8-stage methodology: (1) data preparation, (2) exploratory data analysis, (3) panel diagnostics, (4) baseline estimation, (5) composite index model, (6) post-estimation diagnostics, (7) robustness checks, (8) Granger causality / interpretation.

Run cells **top to bottom**. Each stage is self-contained but depends on the `panel` DataFrame built in Stage 1.

## Research Gap and Contribution

*(Draft language for the thesis introduction/literature review — adapt citations and phrasing to your own reading of the literature; the claims below should be checked against your actual literature review rather than taken as pre-verified facts.)*

**Why South Asia?** Existing empirical work on digital transformation and entrepreneurial activity (e.g., Anton 2024 on the EU, Zhang et al. 2022 on a global 101-country sample, Galindo-Martin et al. 2019 on Europe) is concentrated on high- and upper-middle-income regions with mature digital infrastructure and dense, harmonized statistical reporting (the EU's DESI index, for instance, does not exist for South Asia). South Asia is a materially different setting: lower and more heterogeneous levels of digital infrastructure across neighboring countries (from Bhutan/Nepal to India), weaker formal financial systems, and a very different entrepreneurial ecosystem (higher informal-sector share, different regulatory cost structures for business registration). Whether the digital-transformation -> entrepreneurship relationship found in high-income settings replicates in this environment is an open empirical question, not something that can be assumed from EU or global-average estimates.

**Why these ICT variables?** Internet use, mobile subscriptions, fixed broadband, and secure internet servers capture four distinct dimensions of digital transformation — access, connectivity, infrastructure quality, and digital trust/security — following the same logic as the DESI sub-index decomposition used by Anton (2024) and the connectivity/artefact distinction in Zhang et al. (2022), but built from globally available World Bank WDI indicators so the analysis is replicable outside the EU where DESI is not computed.

**What previous studies have not examined:** (i) a South Asia-specific panel spanning multiple digital-transformation dimensions simultaneously (most South Asia-focused work uses a single connectivity proxy); (ii) an explicit panel cointegration test given the mixed order of integration typical of infrastructure vs. macroeconomic series in developing economies; (iii) panel Granger causality to probe the direction of the digital transformation <-> entrepreneurship relationship, which the cross-sectional/short-panel EU literature generally does not test.

**Contribution:** this thesis extends the digital-transformation-entrepreneurship literature to a low/lower-middle-income, high-heterogeneity regional setting, uses a fuller diagnostic pipeline (unit roots, cointegration, cross-sectional dependence, Hausman-guided estimator choice) than is typical in this literature, and explicitly tests — rather than assumes — the direction of causality between digital infrastructure and new business formation.

## Theoretical Foundation

Four established theoretical perspectives motivate the hypothesized relationship between digital transformation and entrepreneurial activity:

- **Schumpeter's Theory of Innovation (1934/1942)** — entrepreneurship is fundamentally driven by "creative destruction": new technologies (including digital infrastructure) create opportunities for new combinations of resources, enabling new entrants to challenge incumbents. Digital transformation, in this view, is a technological shock that expands the opportunity set available to would-be entrepreneurs.
- **Diffusion of Innovation Theory (Rogers, 1962)** — the adoption of a new technology (here, internet, mobile, and broadband infrastructure) follows a predictable diffusion curve across a population; as digital infrastructure diffuses through a country, more individuals cross the threshold where digital tools make starting a business newly viable, which should show up as rising new business density as connectivity indicators rise.
- **Technology Acceptance Model — TAM (Davis, 1989)** — perceived usefulness and ease of use determine whether individuals adopt a technology; better digital infrastructure (faster broadband, more secure services) lowers the effort barrier to using digital tools for business purposes (e-commerce, digital payments, online marketing), which is one mechanism connecting `BBAND`/`SECSERV` to `NBD`.
- **Resource-Based View — RBV (Barney, 1991)** — firms (including nascent ones) compete on the bundle of resources and capabilities they can access; digital infrastructure is increasingly a resource that determines which capabilities (market reach, payment processing, remote collaboration) a new venture can draw on at low cost, which is why access to finance (`CREDIT`) and digital access are modeled jointly rather than digital variables alone.

These four perspectives motivate treating digital transformation as multi-dimensional (Schumpeter/TAM: infrastructure lowers barriers) and as interacting with financial and macroeconomic context (RBV: resources are bundled, not isolated) — which is the basis for the multi-variable specification with financial and macro controls used in Stage 4, rather than a single-indicator model.

## Stage 1 — Data Preparation

Steps performed:
1. Import the WDI Excel export (`Data` sheet).
2. Reshape from wide (one column per year) to long/panel format.
3. Replace World Bank's `".."` missing-value placeholder with `NaN` and convert to numeric.
4. Rename WDI indicator codes to short, readable variable names.
5. Pivot to a `Country x Year` panel.
6. Restrict to the window with the best data coverage across all variables (checked below) and interpolate small remaining within-country gaps.

**Note:** update `DATA_PATH` below if your file is stored elsewhere.

```python
DATA_PATH = "data.xlsx"   # place data.xlsx in the same folder as this notebook, or give a full path

raw = pd.read_excel(DATA_PATH, sheet_name="Data")

# Drop footer/metadata rows (rows without a Series Code)
raw = raw.dropna(subset=["Series Code"]).copy()

print("Raw shape:", raw.shape)
raw.head()
```

```python
# Reshape wide -> long
year_cols = [c for c in raw.columns if c.startswith("20")]

long_df = raw.melt(
    id_vars=["Country Name", "Country Code", "Series Name", "Series Code"],
    value_vars=year_cols,
    var_name="Year",
    value_name="Value",
)
long_df["Year"] = long_df["Year"].str.extract(r"(\d{4})").astype(int)

# World Bank uses ".." for missing values
long_df["Value"] = long_df["Value"].replace("..", np.nan)
long_df["Value"] = pd.to_numeric(long_df["Value"], errors="coerce")

long_df.head()
```

```python
# Map WDI indicator codes to short variable names
code_map = {
    "IC.BUS.NDNS.ZS": "NBD",         # New business density (dependent variable)
    "IT.NET.USER.ZS": "INTERNET",    # Individuals using the Internet (%)
    "IT.CEL.SETS.P2": "MOBILE",      # Mobile cellular subscriptions (per 100 people)
    "IT.NET.BBND.P2": "BBAND",       # Fixed broadband subscriptions (per 100 people)
    "IT.NET.SECR.P6": "SECSERV",     # Secure Internet servers (per 1 million people)
    "NY.GDP.PCAP.CD": "GDPPC",       # GDP per capita, current US$
    "FS.AST.PRVT.GD.ZS": "CREDIT",   # Domestic credit to private sector (% GDP)
    "BX.KLT.DINV.WD.GD.ZS": "FDI",   # FDI net inflows (% GDP)
    "FP.CPI.TOTL.ZG": "INFLATION",   # Inflation, consumer prices (annual %)
    "NY.GDP.MKTP.KD.ZG": "GDPGROWTH",# GDP growth (annual %) -- used only in robustness checks
    "BX.GSR.CCIS.ZS": "ICTEXPORT",   # ICT service exports -- excluded from main model (see notes)
}
long_df["Var"] = long_df["Series Code"].map(code_map)

panel_full = long_df.pivot_table(index=["Country Name", "Year"], columns="Var", values="Value").reset_index()
panel_full = panel_full.rename(columns={"Country Name": "Country"}).sort_values(["Country", "Year"]).reset_index(drop=True)

print("Full panel (all available years):", panel_full.shape)
panel_full.head()
```

```python
# Check data coverage by year to choose the estimation window
core_vars = ["NBD","INTERNET","MOBILE","BBAND","SECSERV","GDPPC","CREDIT","FDI","INFLATION"]
coverage = panel_full.groupby("Year")[core_vars].apply(lambda d: d.notna().mean().round(2))
coverage
```

```python
# Restrict to 2010-2023: the widest window with complete or near-complete coverage
# (Secure Internet Servers only starts in 2010; 2024-2025 are largely unreported for several series)
panel = panel_full[(panel_full["Year"] >= 2010) & (panel_full["Year"] <= 2023)].reset_index(drop=True)
panel = panel.sort_values(["Country", "Year"]).reset_index(drop=True)

print("Restricted panel shape:", panel.shape)
print("\nMissing values before interpolation:")
print(panel[core_vars + ["GDPGROWTH"]].isna().sum())
```

```python
# Interpolate small within-country gaps (linear), justified since these are slow-moving
# macroeconomic/infrastructure series with no structural breaks in the missing years.
all_vars = core_vars + ["GDPGROWTH", "ICTEXPORT"]
for v in all_vars:
    panel[v] = panel.groupby("Country")[v].transform(lambda s: s.interpolate(limit_direction="both"))

print("Missing values after interpolation:")
print(panel[all_vars].isna().sum())
print("\nFinal balanced panel:", panel.shape, "= 6 countries x 14 years")
panel.head(10)
```

## Stage 3 — Panel Diagnostics

1. **Multicollinearity** — Variance Inflation Factors (VIF); values below 10 (ideally below 5) indicate multicollinearity is not a serious concern.
2. **Cross-sectional dependence** — Pesaran CD test. H0: no cross-sectional dependence.
3. **Panel unit roots** — Fisher-type test (combination of per-country Augmented Dickey-Fuller p-values via the Maddala-Wu chi-square statistic). This is a practical Python approximation of the IPS/CIPS tests referenced in the literature, appropriate for a small-N panel like this one. H0: all panels contain a unit root.


```python
# --- Multicollinearity: VIF ---
X_vars = ["INTERNET","MOBILE","BBAND","SECSERV","GDPPC","CREDIT","FDI","INFLATION"]
X_indep = add_constant(panel[X_vars])

vif = pd.Series(
    [variance_inflation_factor(X_indep.values, i) for i in range(X_indep.shape[1])],
    index=X_indep.columns,
)
print("Variance Inflation Factors:")
vif.round(2)
```

```python
# --- Cross-sectional dependence: Pesaran CD test ---
def pesaran_cd(df, var, entity_col="Country", time_col="Year"):
    wide = df.pivot(index=time_col, columns=entity_col, values=var)
    N, T = wide.shape[1], wide.shape[0]
    corr_mat = wide.corr().values
    rho_sum = sum(corr_mat[i, j] for i in range(N) for j in range(i + 1, N))
    CD = np.sqrt(2 * T / (N * (N - 1))) * rho_sum
    p_value = 2 * (1 - stats.norm.cdf(abs(CD)))
    return CD, p_value

print("Pesaran CD test (H0: no cross-sectional dependence)\n")
for v in desc_vars:
    cd, p = pesaran_cd(panel, v)
    flag = "**" if p < 0.05 else ""
    print(f"{v:10s}  CD = {cd:7.3f}   p = {p:.3f} {flag}")
```

```python
# --- Panel unit root: Fisher-type (Maddala-Wu) combination of per-country ADF tests ---
def fisher_panel_unit_root(df, var, entity_col="Country", time_col="Year", regression="ct"):
    pvals = []
    for c, g in df.groupby(entity_col):
        series = g.sort_values(time_col)[var].dropna().values
        try:
            _, p, *_ = adfuller(series, regression=regression, autolag="AIC")
            pvals.append(np.clip(p, 1e-6, 1 - 1e-6))
        except Exception:
            continue
    pvals = np.array(pvals)
    stat = -2 * np.sum(np.log(pvals))
    dof = 2 * len(pvals)
    p_combined = 1 - stats.chi2.cdf(stat, dof)
    return stat, p_combined, pvals

print("Fisher-type panel unit root test - LEVELS (trend + constant)")
print("H0: all panels contain a unit root\n")
level_summary = {}
for v in desc_vars:
    stat, p, pvals = fisher_panel_unit_root(panel, v, regression="ct")
    level_summary[v] = (stat, p)
    print(f"{v:10s}  chi2 = {stat:7.2f}   p = {p:.3f}   (country p-values: {np.round(pvals,3)})")
```

```python
# First-differenced series (to check order of integration)
panel_d = panel.sort_values(["Country", "Year"]).copy()
for v in desc_vars:
    panel_d["d_" + v] = panel_d.groupby("Country")[v].diff()

print("Fisher-type panel unit root test - FIRST DIFFERENCES (constant only)\n")
diff_summary = {}
for v in desc_vars:
    dsub = panel_d.dropna(subset=["d_" + v])
    stat, p, _ = fisher_panel_unit_root(dsub, "d_" + v, regression="c")
    diff_summary[v] = (stat, p)
    print(f"d_{v:8s}  chi2 = {stat:7.2f}   p = {p:.3f}")

print("\nInterpretation: variables that reject the unit-root null only after differencing are")
print("integrated of order I(1); variables that reject it in levels are I(0). Mixed orders of")
print("integration across regressors motivate cross-checking the FE/RE results against a")
print("long-run (ARDL-type) specification -- see the Interpretation section at the end.")
```

### Stage 3d — Panel Cointegration Test

The unit root tests above show **mixed orders of integration**: some series (e.g. `NBD`, `GDPPC`, `FDI`) reject a unit root in levels, while others (`INTERNET`, `MOBILE`, `BBAND`, `SECSERV`, `CREDIT`, `INFLATION`) only become stationary after first-differencing, i.e. I(1). Whenever regressors are a mix of I(0) and I(1), a long-run relationship between them can still be meaningful, but it must be tested for **cointegration** rather than assumed — this was the biggest methodological gap in the previous version of this notebook.

Two complementary residual-based tests are implemented, since dedicated Pedroni/Kao/Westerlund routines are not available in the standard Python econometrics stack (they exist in Stata's `xtcointtest`, EViews, and R's `plm`/`pco` packages, which are worth cross-checking against for the final thesis submission):

1. **Pedroni-style (heterogeneous) test** — estimates a separate long-run cointegrating regression for *each country*, tests each country's residuals for a unit root (ADF), and combines the six p-values with the Fisher (Maddala-Wu) chi-square statistic. This allows the long-run relationship to differ across countries, matching the spirit of Pedroni's group-mean statistics.
2. **Kao-style (homogeneous) test** — uses the residuals from the single **pooled two-way Fixed Effects** regression (one common long-run relationship for all countries), splits them by country, and combines the per-country ADF p-values the same way. This matches the spirit of Kao's pooled/common-slope approach.

**H0 for both tests: no cointegration (the regression residuals contain a unit root).** Rejecting H0 supports the existence of a long-run equilibrium relationship between digital transformation and entrepreneurial activity.

**Caveat for the thesis write-up:** because the Fisher combination in both tests uses standard Dickey-Fuller/chi-square critical values rather than the exact Pedroni/Kao asymptotic distributions (which are simulated and tabulated separately, and rely on N -> infinity asymptotics not well suited to a 6-country panel), treat the results as **indicative** rather than definitive. Report this limitation explicitly alongside the results.

```python
# --- Pedroni-style panel cointegration test (heterogeneous, country-by-country) ---
# NOTE ON DEGREES OF FREEDOM: each country only has 14 annual observations. Regressing NBD on
# all 8 explanatory variables country-by-country leaves just 14 - 9 = 5 residual degrees of
# freedom per country, which risks overfitting the long-run regression and producing spuriously
# extreme (near-zero) ADF p-values on the residuals. p-values are clipped away from exactly 0/1
# to keep the Fisher combination finite; treat this test as a coarse screen, not a precise
# p-value, and prefer the more parsimonious Kao-style test (pooled slope, one long-run vector)
# as the primary cointegration result for the thesis.
def pedroni_style_cointegration(df, y_var, x_vars, entity_col="Country", time_col="Year"):
    pvals = []
    country_results = {}
    for c, g in df.groupby(entity_col):
        g = g.sort_values(time_col)
        Xc = sm.add_constant(g[x_vars])
        yc = g[y_var]
        ols_c = sm.OLS(yc, Xc).fit()
        resid_c = ols_c.resid.values
        try:
            _, p, *_ = adfuller(resid_c, regression="c", autolag="AIC")
        except Exception:
            continue
        p = np.clip(p, 1e-6, 1 - 1e-6)
        pvals.append(p)
        country_results[c] = p
    pvals = np.array(pvals)
    stat = -2 * np.sum(np.log(pvals))
    dof = 2 * len(pvals)
    p_combined = 1 - stats.chi2.cdf(stat, dof)
    return stat, p_combined, country_results

ped_stat, ped_p, ped_country = pedroni_style_cointegration(panel, "NBD", X_vars)
print("Pedroni-style panel cointegration test (heterogeneous long-run vectors)")
print(f"chi2 = {ped_stat:.2f},  p = {ped_p:.4f}")
print("Country-level residual ADF p-values (clipped away from 0/1):", {k: round(v, 4) for k, v in ped_country.items()})
print("->", "Reject H0: evidence of cointegration" if ped_p < 0.05 else "Fail to reject H0: no strong evidence of cointegration")
print("\nCaveat: only 5 residual degrees of freedom per country-level regression (14 obs, 9 params) --")
print("this heterogeneous test is much less reliable here than the pooled Kao-style test below.")
```

```python
def kao_style_cointegration(resid_series, entity_col="Country", time_col="Year"):
    df_r = resid_series.reset_index()
    df_r.columns = [entity_col, time_col, "resid"]
    pvals = []
    for c, g in df_r.groupby(entity_col):
        series = g.sort_values(time_col)["resid"].values
        try:
            _, p, *_ = adfuller(series, regression="n", autolag="AIC")
            pvals.append(np.clip(p, 1e-6, 1 - 1e-6))
        except Exception:
            continue
    pvals = np.array(pvals)
    stat = -2 * np.sum(np.log(pvals))
    dof = 2 * len(pvals)
    p_combined = 1 - stats.chi2.cdf(stat, dof)
    return stat, p_combined

print("Kao-style test function defined -- it will be applied to the two-way FE residuals right after Stage 4 estimates the FE model below.")
```

## Stage 4 — Baseline Estimation

Three standard panel estimators, followed by the Hausman specification test to choose between Fixed Effects (FE) and Random Effects (RE).

- **Pooled OLS** — ignores country heterogeneity (baseline reference only).
- **Fixed Effects (FE)** — controls for time-invariant country characteristics (entity effects) and common shocks (time effects).
- **Random Effects (RE)** — treats country effects as random and uncorrelated with the regressors.
- **Hausman test** — H0: RE is consistent and efficient (country effects uncorrelated with regressors). Rejecting H0 favors FE.

Standard errors are clustered by country throughout.

```python
pdf = panel.set_index(["Country", "Year"])
y = pdf["NBD"]
X = add_constant(pdf[X_vars])

pooled = PooledOLS(y, X).fit(cov_type="clustered", cluster_entity=True)
fe = PanelOLS(y, X, entity_effects=True, time_effects=True, drop_absorbed=True).fit(
    cov_type="clustered", cluster_entity=True
)
re = RandomEffects(y, X).fit(cov_type="clustered", cluster_entity=True)

print(pooled.summary)
```

```python
print(fe.summary)

# Explicit confirmation that BOTH entity and time effects are estimated, matching the
# two-way FE model in the equation above (mu_i for entity effects, lambda_t for time effects).
print("\nIncluded effects in this FE model:", fe.included_effects)
assert set(fe.included_effects) == {"Entity", "Time"}, "Two-way FE not estimated as specified -- check entity_effects/time_effects flags."
print("Confirmed: this is a two-way (entity + time) Fixed Effects model, matching mu_i + lambda_t in the model equation.")
```

```python
print(re.summary)
```

```python
# --- Hausman test (manual implementation; not built into linearmodels) ---
def hausman(fe_res, re_res):
    common = [c for c in fe_res.params.index if c in re_res.params.index and c != "const"]
    b_diff = fe_res.params[common] - re_res.params[common]
    v_diff = fe_res.cov.loc[common, common] - re_res.cov.loc[common, common]
    stat = b_diff.values @ np.linalg.pinv(v_diff.values) @ b_diff.values
    dof = len(common)
    p = 1 - stats.chi2.cdf(stat, dof)
    return stat, dof, p

h_stat, h_dof, h_p = hausman(fe, re)
print(f"Hausman test:  chi2({h_dof}) = {h_stat:.3f},  p = {h_p:.4f}")
print("Decision:", "Reject RE -> Fixed Effects preferred" if h_p < 0.05 else "Fail to reject -> Random Effects not rejected")
```

```python
# --- Kao-style panel cointegration test (homogeneous, pooled two-way FE residuals) ---
# Uses the FE model's residuals from the two-way Fixed Effects model estimated above (Stage 4).
kao_stat, kao_p = kao_style_cointegration(fe.resids)
print("Kao-style panel cointegration test (homogeneous long-run vector, pooled two-way FE residuals)")
print(f"chi2 = {kao_stat:.2f},  p = {kao_p:.4f}")
print("->", "Reject H0: evidence of cointegration" if kao_p < 0.05 else "Fail to reject H0: no strong evidence of cointegration")
print("\n(See Stage 3d above for the Pedroni-style heterogeneous version and full caveats on both tests.)")
```

## Stage 5 — Composite ICT Index (robustness / alternative specification)

The four digital-transformation indicators (`INTERNET`, `MOBILE`, `BBAND`, `SECSERV`) are correlated dimensions of the same underlying construct. As a complementary specification, a single **ICT Index** is built via first-component PCA on the standardized variables, and the FE model is re-estimated with this composite regressor in place of the four individual indicators.

```python
ict_vars = ["INTERNET","MOBILE","BBAND","SECSERV"]
scaler = StandardScaler()
Z = scaler.fit_transform(panel[ict_vars])

pca = PCA(n_components=1)
panel["ICT_INDEX"] = pca.fit_transform(Z).flatten()

print(f"PC1 explained variance ratio: {pca.explained_variance_ratio_[0]:.3f}")
print("Loadings (higher = more weight on that indicator):")
pd.Series(pca.components_[0], index=ict_vars).round(3)
```

```python
pdf2 = panel.set_index(["Country", "Year"])
y2 = pdf2["NBD"]
X2 = add_constant(pdf2[["ICT_INDEX", "GDPPC", "CREDIT", "FDI", "INFLATION"]])

fe_ict = PanelOLS(y2, X2, entity_effects=True, time_effects=True, drop_absorbed=True).fit(
    cov_type="clustered", cluster_entity=True
)
print(fe_ict.summary)
```

## Stage 6 — Post-Estimation Diagnostics (on the baseline FE model)

- **Serial correlation** — AR(1) test on FE residuals (clustered by country).
- **Heteroskedasticity** — Breusch-Pagan test of squared residuals on fitted values.
- **Cross-sectional dependence** — Pesaran CD test applied to the FE residuals.

```python
resid_df = fe.resids.reset_index()
resid_df.columns = ["Country", "Year", "resid"]
resid_df = resid_df.sort_values(["Country", "Year"])
resid_df["resid_lag"] = resid_df.groupby("Country")["resid"].shift(1)

sc_data = resid_df.dropna()
sc_X = sm.add_constant(sc_data["resid_lag"])
sc_model = sm.OLS(sc_data["resid"], sc_X).fit(
    cov_type="cluster", cov_kwds={"groups": sc_data["Country"]}
)
print("Serial correlation check: AR(1) coefficient on FE residuals")
print(f"rho = {sc_model.params['resid_lag']:.3f},  p = {sc_model.pvalues['resid_lag']:.4f}")
print("->", "Evidence of serial correlation" if sc_model.pvalues["resid_lag"] < 0.05 else "No strong evidence of serial correlation")
```

```python
fitted = fe.fitted_values.values.flatten()
resid_vals = fe.resids.values
bp_X = sm.add_constant(fitted)
bp_aux = sm.OLS(resid_vals ** 2, bp_X).fit()
lm_stat = bp_aux.rsquared * len(resid_vals)
lm_p = 1 - stats.chi2.cdf(lm_stat, 1)

print("Breusch-Pagan heteroskedasticity test")
print(f"LM = {lm_stat:.3f},  p = {lm_p:.4f}")
print("->", "Evidence of heteroskedasticity (clustered SEs already used in FE model)" if lm_p < 0.05 else "No strong evidence of heteroskedasticity")
```

```python
resid_wide = resid_df.pivot(index="Year", columns="Country", values="resid")
N, T = resid_wide.shape[1], resid_wide.shape[0]
c_mat = resid_wide.corr().values
rho_sum = sum(c_mat[i, j] for i in range(N) for j in range(i + 1, N))
CD_resid = np.sqrt(2 * T / (N * (N - 1))) * rho_sum
p_resid = 2 * (1 - stats.norm.cdf(abs(CD_resid)))

print("Post-estimation Pesaran CD test on FE residuals")
print(f"CD = {CD_resid:.3f},  p = {p_resid:.4f}")
print("->", "Residual cross-sectional dependence remains -- consider Driscoll-Kraay SEs as a further robustness step"
      if p_resid < 0.05 else "No remaining cross-sectional dependence in residuals")
```

## Stage 7 — Robustness Checks

(a) Replace GDP per capita with GDP growth as the macroeconomic control.
(b) Drop each digital-transformation variable one at a time.
(c) Lag all digital-transformation variables by one year (addresses potential simultaneity / reverse causality).

```python
# (a) Alternative control variable: GDP growth instead of GDP per capita
Xa = add_constant(pdf[["INTERNET","MOBILE","BBAND","SECSERV","GDPGROWTH","CREDIT","FDI","INFLATION"]]) \
     if "GDPGROWTH" in pdf.columns else None
Xa = add_constant(panel.set_index(["Country","Year"])[["INTERNET","MOBILE","BBAND","SECSERV","GDPGROWTH","CREDIT","FDI","INFLATION"]])

fe_a = PanelOLS(y, Xa, entity_effects=True, time_effects=True, drop_absorbed=True).fit(
    cov_type="clustered", cluster_entity=True
)
print("(a) GDP growth instead of GDP per capita:\n")
print(fe_a.summary)
```

```python
# (b) Drop one digital variable at a time
print("(b) Drop-one-digital-variable-at-a-time\n")
digital_vars = ["INTERNET","MOBILE","BBAND","SECSERV"]
drop_results = {}
for drop_v in digital_vars:
    remaining = [v for v in X_vars if v != drop_v]
    Xb = add_constant(pdf[remaining])
    fe_b = PanelOLS(y, Xb, entity_effects=True, time_effects=True, drop_absorbed=True).fit(
        cov_type="clustered", cluster_entity=True
    )
    drop_results[drop_v] = fe_b
    print(f"Dropped {drop_v:10s} | R2 = {fe_b.rsquared:.3f} | coefficients:")
    print(fe_b.params.round(4).to_string())
    print()
```

```python
# (c) Lag digital-transformation variables by one year
panel_lag = panel.sort_values(["Country","Year"]).copy()
for v in digital_vars:
    panel_lag["L_" + v] = panel_lag.groupby("Country")[v].shift(1)

panel_lag2 = panel_lag.dropna(subset=["L_" + v for v in digital_vars])
pdf_lag = panel_lag2.set_index(["Country","Year"])
y_lag = pdf_lag["NBD"]
X_lag = add_constant(pdf_lag[["L_INTERNET","L_MOBILE","L_BBAND","L_SECSERV","GDPPC","CREDIT","FDI","INFLATION"]])

fe_lag = PanelOLS(y_lag, X_lag, entity_effects=True, time_effects=True, drop_absorbed=True).fit(
    cov_type="clustered", cluster_entity=True
)
print("(c) Digital-transformation variables lagged by one year:\n")
print(fe_lag.summary)
```

**Why these robustness checks, and what they show:**

- **(a) GDP growth instead of GDP per capita** — GDP per capita is a *level* (persistent, slow-moving) while GDP growth is a *rate* (year-to-year volatility). If the digital-transformation coefficients flipped sign or lost significance under this swap, it would suggest the baseline result is an artifact of how the macroeconomic control is specified rather than a genuine digital-transformation effect. In this run, `MOBILE` and `BBAND` retain the same sign and remain the two variables carrying the estimated effect, which supports the baseline story.
- **(b) Drop-one-digital-variable-at-a-time** — because `INTERNET` and `SECSERV` are highly correlated (see the correlation matrix, Stage 2), removing one at a time checks whether the results are being driven by collinearity rather than by each variable's own information content. R² barely moves when `INTERNET` or `SECSERV` is dropped (0.826-0.825 vs. 0.826 baseline), but drops noticeably when `BBAND` is removed (0.682) — evidence that fixed broadband is doing most of the explanatory work among the four digital indicators, while internet-use and secure-servers largely proxy for the same underlying variation.
- **(c) One-year lag of digital variables** — regressing `NBD` on *last year's* digital-transformation indicators is a partial (not full) safeguard against reverse causality: if new firms mechanically raised this year's internet/mobile/broadband uptake, that channel is shut off by using lagged regressors. The lagged coefficients stay close in sign, magnitude, and significance to the contemporaneous FE model, which is reassuring but does **not** rule out reverse causality outright (see the endogeneity discussion after Stage 8).

**Overall:** the qualitative pattern — connectivity/infrastructure variables (mobile, broadband) positively associated with new business density, GDP per capita and inflation negatively associated, credit access positively associated — is stable across all three checks. This increases confidence that the baseline FE estimates are not an artifact of a specific control-variable choice, a specific set of regressors, or the timing assumption of contemporaneous effects.

## Stage 8 — Panel Granger Causality

Tests whether past values of each digital-transformation variable help predict `NBD` beyond its own past values, and vice versa. Implemented as a **Dumitrescu-Hurlin-style** average Wald statistic across countries (per-country Granger tests combined into a panel `Z-bar` statistic).

**Caveat:** with only 6 cross-sectional units this is indicative rather than a fully powered asymptotic test — the Dumitrescu-Hurlin test is designed for larger N. Report it as suggestive evidence of lead-lag patterns, not as definitive causal proof.

```python
def panel_granger(df, x_var, y_var, entity_col="Country", time_col="Year", maxlag=1):
    wstats = []
    for c, g in df.groupby(entity_col):
        g = g.sort_values(time_col)
        data = g[[y_var, x_var]].dropna()
        if len(data) < maxlag + 3:
            continue
        try:
            res = grangercausalitytests(data, maxlag=maxlag, verbose=False)
            wstats.append(res[maxlag][0]["ssr_ftest"][0])
        except Exception:
            continue
    wstats = np.array(wstats)
    N, k = len(wstats), maxlag
    W_bar = wstats.mean()
    Z_bar = np.sqrt(N / (2 * k)) * (W_bar - k)
    p_z = 2 * (1 - stats.norm.cdf(abs(Z_bar)))
    return W_bar, Z_bar, p_z

print("Panel Granger causality (maxlag = 1)\n")
test_vars = digital_vars + ["ICT_INDEX"]
for v in test_vars:
    w1, z1, p1 = panel_granger(panel, v, "NBD")
    w2, z2, p2 = panel_granger(panel, "NBD", v)
    print(f"{v + ' -> NBD':20s} W_bar={w1:6.3f}  Z_bar={z1:6.3f}  p={p1:.3f}")
    print(f"{'NBD -> ' + v:20s} W_bar={w2:6.3f}  Z_bar={z2:6.3f}  p={p2:.3f}")
    print()
```

### Endogeneity and Reverse Causality — Discussion

The Granger causality results above are suggestive of a genuine endogeneity concern that should be stated explicitly as a **limitation**, not resolved away:

- For `INTERNET` and `MOBILE`, causality runs **in both directions** (each Granger-causes `NBD`, and `NBD` also Granger-causes each of them). This is exactly the pattern predicted by theory on both sides: digital access lowers barriers to starting a business (the hypothesis this thesis tests), but a more active entrepreneurial sector also *drives demand* for internet and mobile connectivity (firms adopting digital tools, e-commerce, digital payments), and governments/telecoms may expand infrastructure faster in more economically dynamic markets. Both channels can be operating simultaneously.
- For the composite `ICT_INDEX`, causality is **unidirectional** (ICT_INDEX -> NBD is significant; NBD -> ICT_INDEX is not), which is more consistent with digital infrastructure being a genuine driver rather than a passive by-product of entrepreneurial activity — but this single result should not be over-read given the caveats below.
- **Why this matters for interpretation:** if reverse causality (or a third factor jointly driving both, such as an unobserved country-specific reform push) is present, the FE coefficients in Stage 4 are not cleanly interpretable as the causal effect of digital transformation on entrepreneurship — they capture an equilibrium association between the two, conditional on the fixed effects and controls included.
- **What was done to partially address this:** (i) two-way fixed effects remove time-invariant country characteristics and common shocks that could jointly drive both variables; (ii) the lagged-regressor robustness check (Stage 7c) reduces same-year simultaneity; (iii) Granger causality (Stage 8) was used to check the direction of lead-lag patterns rather than assuming exogeneity.
- **What was not done, and should be flagged as a limitation:** no instrumental-variable (IV) or System-GMM (Arellano-Bond/Blundell-Bond) estimation was performed. These would be the standard next step to obtain causally-identified estimates, but they require either a credible instrument for digital transformation (e.g., historical telecom infrastructure, distance to submarine cables, staggered rollout policies) or a dynamic panel with enough time periods per country to support GMM's internal instruments — both of which are constrained by this dataset's 6-country, 14-year structure. This should be stated in the thesis as a direction for future research rather than something this notebook resolves.

- `INTERNET` and `SECSERV` are not statistically significant in the FE model, even though they showed positive raw correlations with `NBD` (Stage 2) and were significant in the naive Pooled OLS/RE specifications — this is a classic sign that part of their apparent effect in the pooled models was actually cross-country heterogeneity (e.g., generally more digitized countries also have more new-firm registration for unrelated structural reasons), which the FE model correctly nets out.
- `GDPPC` (-) and `INFLATION` (-) are negative and significant; `CREDIT` (+) is positive and significant; `FDI` is not significant at conventional levels. - The **composite ICT Index model (Stage 5)** — a single PC1 index combining all four digital variables — is positive and highly significant (coefficient = 0.106, p < 0.001), reinforcing that digital transformation *as a bundle* is positively associated with entrepreneurial activity, even where the individual indicators are not all separately significant once collinearity between them is accounted for.

**3. Diagnostics (Stage 6).** No strong evidence of serial correlation (AR(1) p = 0.87) or heteroskedasticity (Breusch-Pagan p = 0.07) in the FE residuals. Residual cross-sectional dependence is still present (Pesaran CD p = 0.011) even after two-way FE — country-clustered standard errors partially address this, but Driscoll-Kraay standard errors (robust to both heteroskedasticity and cross-sectional dependence) would be a natural next robustness step if reviewers push on this point.

**4. Cointegration (Stage 3d).** Both the Pedroni-style (heterogeneous) and Kao-style (homogeneous, pooled FE-residual) tests should be read together with the mixed unit-root results: report the actual chi2/p-values produced when you run the notebook, and note explicitly that the Fisher-combination approach used here is an approximation of the true Pedroni/Kao asymptotics (see the caveat in Stage 3d) — cross-checking in Stata or R is recommended before treating the long-run relationship as confirmed.

**5. Robustness (Stage 7).** The qualitative pattern (connectivity variables positive, GDP per capita and inflation negative, credit positive) is stable across the alternative control, the drop-one-variable checks, and the one-year lag specification — see the discussion cell directly above for the detailed reasoning.

**6. Causality (Stage 8) and endogeneity.** See the dedicated discussion cell above. In short: causality is bidirectional for `INTERNET`/`MOBILE` and unidirectional (ICT -> NBD) for the composite index, which should be reported as *suggestive* evidence given the small N, not as proof of a one-way causal effect.

**7. Policy implications.** The results point most consistently to **mobile and broadband connectivity infrastructure** and **access to finance (domestic credit)** as the strongest correlates of new business formation in this sample, more so than internet penetration or server security infrastructure in isolation. For South Asian policymakers, this suggests infrastructure investment paired with financial-sector deepening may be a more effective lever for entrepreneurship promotion than digital-access campaigns alone — though, given the endogeneity caveats above, this should be framed as an association consistent with that policy story rather than as a proven causal recommendation.

## Limitations

State these explicitly in the thesis — they materially affect how strongly any conclusion should be worded:

1. **Small sample size.** The panel covers 6 countries over 14 years (84 observations). This is a small-N, moderate-T panel: asymptotic properties that standard panel estimators (and especially the Fisher-type unit-root and cointegration tests used here) rely on are derived for N and/or T -> infinity. With N = 6, results — particularly the panel unit-root and cointegration test p-values — should be read as **indicative, not definitive**, and cross-checked against specialized software (Stata, R, EViews) that implements the exact Pedroni/Kao/Westerlund/IPS asymptotic distributions before being reported as confirmatory in the final thesis.
2. **No causal identification.** As discussed after Stage 8, no instrumental-variable or GMM strategy was used to address potential reverse causality or omitted joint drivers. All FE coefficients should be described as associations conditional on observed controls and fixed effects, not as causal effects, unless a credible identification strategy is added.
3. **Residual cross-sectional dependence.** Stage 6 finds that Pesaran CD remains significant on the FE residuals even after two-way fixed effects, suggesting a common shock or spillover across these six neighboring/regional economies (e.g., regional trade linkages, shared external shocks) that the model does not fully capture. Driscoll-Kraay standard errors are a natural extension if this needs to be addressed more rigorously than country-clustered SEs allow.
4. **Interpolated data.** A small number of missing values (see Stage 1) were linearly interpolated within-country. While the gaps were short and the series slow-moving, this is a data-construction choice that should be disclosed in the methodology section.
5. **Generalizability.** Findings are specific to these six South Asian economies over 2010-2023 and should not be extrapolated to other regions or periods without re-estimation.
